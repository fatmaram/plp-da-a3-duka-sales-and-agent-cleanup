## Task 2: Create the derived columns

```python
df["revenue"] = df["quantity"] * df["unit_price"]
df["month"] = df["order_date"].dt.month_name()
df["weekday"] = df["order_date"].dt.day_name()
df["is_weekend"] = df["order_date"].dt.dayofweek >= 5
df["price_band"] = np.where(df["unit_price"] >= 28000, "Premium", "Value")

print(df[["order_id", "revenue", "month", "weekday", "is_weekend", "price_band"]].head(3).to_string(index=False))
print("Total revenue:", f"{df['revenue'].sum():,}")
```

```python
print(df["month"].value_counts().reindex(month_order).to_string())
```

```python
print(df["weekday"].value_counts().reindex(day_order).to_string())
```

The five columns sit in the table. `revenue` equals quantity times unit price and totals KES 71,792,500. `month` and `weekday` hold names. `is_weekend` is True for Saturday and Sunday and marks 277 orders. `price_band` is Premium when `unit_price` is 28,000 or more. A Tablet costs exactly 28,000 so Tablet and Laptop form the Premium band. Premium covers 435 orders or 43.5% and Value covers 565 orders. March is the busiest month with 186 orders and May the quietest with 153. Friday is the busiest weekday with 166 orders and Saturday the quietest with 132.

## Task 3: Rank branches by revenue and by number of orders

```python
branch = df.groupby("branch").agg(revenue=("revenue", "sum"), orders=("order_id", "count"))
branch["rev_rank"] = branch["revenue"].rank(ascending=False).astype(int)
branch["ord_rank"] = branch["orders"].rank(ascending=False).astype(int)
branch = branch.sort_values("revenue", ascending=False)
print(branch.rename_axis(None).to_string())
print("Same ranking:", bool((branch["rev_rank"] == branch["ord_rank"]).all()))
```

```python
shares = pd.DataFrame({
    "rev_share": (branch["revenue"] / branch["revenue"].sum() * 100).round(1),
    "order_share": (branch["orders"] / branch["orders"].sum() * 100).round(1),
    "avg_order": (branch["revenue"] / branch["orders"]).round(0).astype(int),
})
print(shares.rename_axis(None).to_string())
print("Company average order:", round(df["revenue"].mean(), 1))
```

**Answer.** The rankings match. Nairobi comes first and Mombasa second and Nakuru third and Kisumu fourth and Eldoret fifth by revenue and by number of orders.

**Why they match.** Average order value sits in a narrow band from KES 68,353 in Kisumu to KES 75,591 in Eldoret. Revenue follows order volume when order size changes little between branches.

**Where they differ.** The shares tell a finer story. Nairobi holds 37.6% of orders and 36.4% of revenue because its average order of KES 69,431 sits below the company average of KES 71,793. Mombasa holds 18.6% of orders and 19.5% of revenue because its average order of KES 75,289 sits above it. Nakuru and Kisumu are only 3 orders apart (162 and 159). Nakuru still leads by KES 1,102,300 in revenue because its average order is about KES 5,500 larger.

**Business reading.** Nairobi earns KES 26,106,100 and that is 1.9 times Mombasa. Order size is similar everywhere so more orders is the main lever for lifting a branch.

## Task 4: Top three products by revenue and by units sold

```python
prod = df.groupby("product").agg(revenue=("revenue", "sum"), units=("quantity", "sum"), unit_price=("unit_price", "first"))
print("Top 3 by revenue")
print(prod["revenue"].nlargest(3).to_string())
print("Top 3 by units")
print(prod["units"].nlargest(3).to_string())
```

```python
prod["revenue_share_pct"] = (prod["revenue"] / prod["revenue"].sum() * 100).round(1)
print(prod.sort_values("revenue", ascending=False)[["units", "unit_price", "revenue_share_pct"]].rename_axis(None).to_string())
```

**Answer.** Laptop and Tablet and Phone make both top three lists in a different order.

By revenue the list reads Laptop at KES 41,535,000 (57.9%) then Tablet at KES 19,348,000 (26.9%) then Phone at KES 8,880,000 (12.4%). By units the list reads Tablet at 691 then Laptop at 639 then Phone at 592.

**Why the lists differ.** Tablet sells 52 more units than Laptop. A Laptop costs KES 65,000 and a Tablet costs KES 28,000 so the Laptop price is 2.3 times higher. Laptop revenue ends up 2.1 times Tablet revenue even with fewer units. Phone holds third place in both lists at KES 15,000.

**Context.** Headphones and Chargers sell 555 and 535 units and sit close to Phone at 592 units. They earn only 1.9% and 0.9% of revenue because they cost KES 2,500 and KES 1,200. Unit counts alone mislead. Laptops and Tablets together carry 84.8% of revenue.

## Task 5: Weekend orders against weekday orders

```python
wk = df.groupby("is_weekend").agg(orders=("order_id", "count"), revenue=("revenue", "sum"),
                                  avg_order_revenue=("revenue", "mean")).round(0).astype(int)
wk.index = ["Weekday", "Weekend"]
print(wk.to_string())
```

```python
days = pd.date_range(df["order_date"].min(), df["order_date"].max())
weekend_days = int((days.dayofweek >= 5).sum())
weekday_days = len(days) - weekend_days
print("Calendar days: weekday", weekday_days, "| weekend", weekend_days)
print("Orders per day: weekday", round(wk.loc["Weekday", "orders"] / weekday_days, 2),
      "| weekend", round(wk.loc["Weekend", "orders"] / weekend_days, 2))
print("Revenue per day: weekday", round(wk.loc["Weekday", "revenue"] / weekday_days),
      "| weekend", round(wk.loc["Weekend", "revenue"] / weekend_days))
print("Weekend share of orders:", round(wk.loc["Weekend", "orders"] / wk["orders"].sum() * 100, 1),
      "% | share of days:", round(weekend_days / len(days) * 100, 1), "%")
```

**Answer.** Weekend orders number 277 and bring KES 19,936,300 in revenue with an average order of KES 71,972. Weekday orders number 723 and bring KES 51,856,200 with an average order of KES 71,724.

**Reading.** Order size is almost identical because a weekend order is worth only 0.3% more. Volume tells the real story. Weekends make up 28.9% of the 180 calendar days and carry 27.7% of orders. Duka books 5.33 orders per weekend day and 5.65 per weekday. A weekend day earns about KES 383,390 and a weekday earns KES 405,127 so a weekend day earns 5.4% less.

**Business reading.** Weekends trail weekdays slightly per day. A weekend promotion would need to lift order volume because basket size already matches the weekday level.

## Task 6: Best and worst month for revenue

```python
monthly = df.groupby("month")["revenue"].sum().reindex(month_order)
print(monthly.to_string())
best, worst = monthly.idxmax(), monthly.idxmin()
print("Best:", best, monthly.max(), "| Worst:", worst, monthly.min())
print("Best is", round((monthly.max() - monthly.min()) / monthly.min() * 100, 1), "% above worst")
print("Worst is", round((monthly.max() - monthly.min()) / monthly.max() * 100, 1), "% below best")
```

```python
days_in_month = pd.Series(days.month_name()).value_counts().reindex(month_order)
daily = (monthly / days_in_month).round(0).astype(int)
print(daily.to_string())
print("Lowest daily average:", daily.idxmin())
print("Daily gap, best against worst month:", round((daily[best] - daily[worst]) / daily[worst] * 100, 1), "%")
```

**Answer.** April is the best month with KES 13,518,300 and June is the worst with KES 10,903,400. April sits 24.0% above June. Said the other way June sits 19.3% below April.

**Fair comparison.** June holds data for 29 days only. Daily averages remove that effect. April earns KES 450,610 per day and June earns KES 375,979 which leaves April 19.8% ahead. May has the lowest daily average at KES 368,239 so the slowdown runs through May and June.

**Business reading.** Revenue sits near KES 11.8M in January and February and climbs to KES 12.4M in March and KES 13.5M in April. It then falls by KES 2.1M from April to May. Leadership should check what drove April and what changed in May.

## Task 8: Spreadsheet check against pandas

```python
pandas_branch = df.groupby("branch")["revenue"].sum()
pandas_channel = df.groupby("channel")["revenue"].sum()
print(pandas_branch.sort_values(ascending=False).to_string())
print("Total:", pandas_branch.sum())
```

```python
path = "excel/duka_branch_summary.xlsx"
if not os.path.exists(path):
    print("Workbook missing: build excel/duka_branch_summary.xlsx then run this notebook again")
else:
    try:
        summary = pd.read_excel(path, sheet_name="Summary")
        xl_branch = summary.set_index("Branch")["Revenue"]
        raw = pd.read_excel(path, sheet_name="Pivot", header=None)
        head = raw.index[raw[0] == "Row Labels"][0]
        pivot = raw.iloc[head + 1:].copy()
        pivot.columns = list(raw.iloc[head])
        pivot = pivot.set_index("Row Labels")
        ok_branch = bool(np.allclose(xl_branch.reindex(pandas_branch.index), pandas_branch))
        ok_pivot_branch = bool(np.allclose(pivot["Grand Total"].reindex(pandas_branch.index).astype(float), pandas_branch))
        ok_pivot_channel = bool(np.allclose(pivot.loc["Grand Total", pandas_channel.index].astype(float), pandas_channel))
        ok_total = bool(np.isclose(float(pivot.loc["Grand Total", "Grand Total"]), df["revenue"].sum()))
        print("SUMIFS branch totals match pandas:", ok_branch)
        print("PivotTable branch totals match pandas:", ok_pivot_branch)
        print("PivotTable channel totals match pandas:", ok_pivot_channel)
        print("PivotTable grand total matches pandas:", ok_total)
    except Exception as err:
        print("Check stopped:", str(err)[:200])
```

The workbook `excel/duka_branch_summary.xlsx` holds a SUMIFS branch revenue table on the Summary sheet and an XLOOKUP region column on the Sales sheet. The Pivot sheet holds a PivotTable of revenue by branch and channel. The check cell above reads the workbook and compares each branch total and each channel total and the grand total with pandas. Every comparison prints True once the workbook sits in the `excel` folder.

The pandas totals to expect are Nairobi 26,106,100 and Mombasa 14,003,700 and Nakuru 11,970,400 and Kisumu 10,868,100 and Eldoret 8,844,200. The grand total is 71,792,500. Channel totals are In-store 40,969,200 and Online 22,786,100 and Phone order 8,037,200.
